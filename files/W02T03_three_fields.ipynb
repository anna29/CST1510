### Same idea, three fields — worked examples
A `while` loop is for repeating until a signal, not a fixed count. Run each cell.

**AI / Data Science**

```python
batches_left = 3
while batches_left > 0:
    print("pulling a batch...")
    batches_left -= 1
```

**Cyber Security**

```python
log_lines = ["login ok", "login ok", "SENTINEL"]
i = 0
while log_lines[i] != "SENTINEL":
    print("checking:", log_lines[i])
    i += 1
print("reached end of log")
```

**IT**

```python
checks = 0
status = "starting"
while status != "healthy" and checks < 3:
    print("checking server...")
    status = "healthy"
    checks += 1
print("server is", status)
```
