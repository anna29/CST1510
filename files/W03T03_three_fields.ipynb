### Same idea, three fields — worked examples
A function can read a variable from outside, but changing it only ever changes its own local copy — the original is always safe. Run each cell.

**AI / Data Science**

```python
def clean_row(value):
    value = value.strip().upper()
    return value

raw = "  srv-01  "
cleaned = clean_row(raw)
print(raw)
print(cleaned)
```

**Cyber Security**

```python
def flag(ip):
    ip = ip + " (checked)"
    return ip

source_ip = "10.0.0.5"
flagged = flag(source_ip)
print(source_ip)
print(flagged)
```

**IT**

```python
def mark_checked(hostname):
    hostname = hostname + " - OK"
    return hostname

host = "srv-01"
result = mark_checked(host)
print(host)
print(result)
```
