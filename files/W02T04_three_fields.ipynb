### Same idea, three fields — worked examples
Know the count in advance? That is what `for` is for. Run each cell.

**AI / Data Science**

```python
n_rows = 5
for i in range(n_rows):
    print(f"processing row {i + 1}")
```

**Cyber Security**

```python
n_ips = 3
for i in range(n_ips):
    print(f"checking known IP #{i + 1}")
```

**IT**

```python
n_servers = 4
for i in range(n_servers):
    print(f"pinging server {i + 1}")
```
