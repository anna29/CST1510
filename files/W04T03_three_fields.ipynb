### Same idea, three fields — worked examples
Storage that outlives the program, read back a line at a time. Run each cell.

**AI / Data Science**

```python
with open("readings.csv", "w") as f:
    f.write("21.4\n")
    f.write("21.9\n")
    f.write("22.0\n")

with open("readings.csv") as f:
    readings = [float(line.strip()) for line in f]
print(readings)
```

**Cyber Security**

```python
with open("access.log", "w") as f:
    f.write("10.0.0.5,FAIL\n")
    f.write("10.0.0.9,OK\n")

with open("access.log") as f:
    for line in f:
        ip, result = line.strip().split(",")
        print(f"{ip}: {result}")
```

**IT**

```python
with open("hosts.cfg", "w") as f:
    f.write("srv-01\n")
    f.write("srv-02\n")

with open("hosts.cfg") as f:
    for line in f:
        print(f"checking {line.strip()}")
```
