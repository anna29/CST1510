### Same idea, three fields — worked examples
Same method mechanism — `self`, called as `object.method()` — three different domains. Run each cell.

**AI / Data Science** — one row of a dataset, checked for quality.

```python
class DatasetRow:
    def __init__(self, row_id, value, missing_fields):
        self.row_id = row_id
        self.value = value
        self.missing_fields = missing_fields

    def summary(self):
        if self.missing_fields > 0:
            return f"{self.row_id}: INCOMPLETE ({self.missing_fields} missing fields)"
        return f"{self.row_id}: OK (value={self.value})"

row = DatasetRow("row_042", 87.5, 0)
print(row.summary())
```

**Cyber Security** — one alert, escalated by failed login count.

```python
class Alert:
    def __init__(self, source_ip, failed_logins):
        self.source_ip = source_ip
        self.failed_logins = failed_logins

    def escalate(self):
        if self.failed_logins >= 10:
            return f"{self.source_ip}: ACTIVE INCIDENT"
        elif self.failed_logins >= 5:
            return f"{self.source_ip}: SUSPICIOUS"
        return f"{self.source_ip}: NORMAL"

a = Alert("192.168.1.14", 7)
print(a.escalate())
```

**IT** — one server, checked for disk usage.

```python
class Server:
    def __init__(self, hostname, disk_used_gb, disk_total_gb):
        self.hostname = hostname
        self.disk_used_gb = disk_used_gb
        self.disk_total_gb = disk_total_gb

    def check_disk(self):
        percent = (self.disk_used_gb / self.disk_total_gb) * 100
        if percent >= 90:
            return f"{self.hostname}: CRITICAL ({percent:.0f}% used)"
        elif percent >= 75:
            return f"{self.hostname}: WARNING ({percent:.0f}% used)"
        return f"{self.hostname}: OK ({percent:.0f}% used)"

s = Server("web1", 82, 100)
print(s.check_disk())
```
