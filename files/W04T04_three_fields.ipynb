### Same idea, three fields — worked examples
The same list-of-dicts shape, wherever the rows come from. Run each cell.

**AI / Data Science**

```python
survey_rows = [
    {"respondent": "P-01", "score": 4},
    {"respondent": "P-02", "score": 5},
]
for row in survey_rows:
    print(row)
```

**Cyber Security**

```python
log_entries = [
    {"ip": "10.0.0.5", "time": "09:14", "result": "FAIL"},
    {"ip": "10.0.0.9", "time": "09:16", "result": "OK"},
]
for entry in log_entries:
    print(f"{entry['time']} {entry['ip']}: {entry['result']}")
```

**IT**

```python
servers = [
    {"hostname": "srv-01", "used": 87, "total": 120},
    {"hostname": "srv-02", "used": 60, "total": 100},
]
for s in servers:
    percent = (s["used"] / s["total"]) * 100
    print(f"{s['hostname']}: {percent:.0f}% used")
```
