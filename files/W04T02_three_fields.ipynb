### Same idea, three fields — worked examples
One shape, three different pieces of information stored by name. Run each cell.

**AI / Data Science**

```python
survey = {"rows": 1200, "missing": 13, "source": "survey_2026"}
for key, value in survey.items():
    print(f"{key}: {value}")
```

**Cyber Security**

```python
alert = {"ip": "10.0.0.5", "failed": 12, "locked": True}
for key, value in alert.items():
    print(f"{key}: {value}")
```

**IT**

```python
server = {"hostname": "srv-01", "used_gb": 87, "total_gb": 120}
for key, value in server.items():
    print(f"{key}: {value}")
```
