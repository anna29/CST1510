### Same idea, three fields — worked examples
Almost every calculation you write in year one is *(part / total) * 100*. Run each cell.

**AI / Data Science** — percentage of rows missing a value

```python
missing_rows = 34
total_rows = 1200
percent_missing = (missing_rows / total_rows) * 100
print(f"{percent_missing:.1f}% of rows missing a value")
```

**Cyber Security** — percentage of login attempts that failed

```python
failed_logins = 18
total_logins = 240
percent_failed = (failed_logins / total_logins) * 100
print(f"{percent_failed:.1f}% of login attempts failed")
```

**IT** — percentage of disk space used

```python
used_gb = 87
total_gb = 120
percent_used = (used_gb / total_gb) * 100
print(f"{percent_used:.1f}% of disk used")
```
