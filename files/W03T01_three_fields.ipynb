### Same idea, three fields — worked examples
One job, reused wherever it's needed — different job, same shape. Run each cell.

**AI / Data Science**

```python
def clean(row):
    return row.strip().upper()

print(clean("  srv-01  "))
print(clean("  srv-02  "))
```

**Cyber Security**

```python
def is_suspicious(ip):
    return ip.startswith("10.0.0.9")

print(is_suspicious("10.0.0.9"))
print(is_suspicious("10.0.0.5"))
```

**IT**

```python
def ping(host):
    return f"{host}: OK"

print(ping("srv-01"))
print(ping("srv-02"))
```
