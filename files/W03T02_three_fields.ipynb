### Same idea, three fields — worked examples
More than one input, and not every input needs to be given every time. Run each cell.

**AI / Data Science**

```python
def load(path, header=True):
    return f"Loading {path} (header={header})"

print(load("data.csv"))
print(load("data.csv", header=False))
```

**Cyber Security**

```python
def scan(target, timeout=5):
    return f"Scanning {target} (timeout={timeout}s)"

print(scan("10.0.0.5"))
print(scan("10.0.0.5", timeout=15))
```

**IT**

```python
def backup(source, destination="/backups"):
    return f"Backing up {source} to {destination}"

print(backup("srv-01"))
print(backup("srv-01", destination="/mnt/offsite"))
```
